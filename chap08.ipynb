```python
import datetime as dt

now = dt.datetime.now()
ampm = "오전" if now.hour < 12 else "오후"
hour = now.hour if now.hour <= 12 else now.hour - 12

print(f"오늘의 날짜 : {now.year}년 {now.month}월 {now.day}일")
print(f"현재시간 : {ampm} {hour}시 {now.minute}분 {now.second}초")
```

```python
from datetime import datetime, timedelta

input_data = input("처음으로 사귄 연도와 월, 일을 입력하시오 : ")
year, month, day = map(int, input_data.split())

start_date = datetime(year, month, day)
anniversary_100 = start_date + timedelta(days=100)

print(f"100일 기념일은 : {anniversary_100.year}년 {anniversary_100.month}월 {anniversary_100.day}일입니다.")
```

```python
import math

for i in range(2, 11):
    result = math.pow(4, i)
    print(f"4**{i:2d} = {result:>10.1f}")
```

```python
import math

for degree in range(0, 181, 10):
    radian = math.radians(degree)
    print(f"{degree:3d} degree = {radian:.3f} radian")
```

```python
import math

for degree in range(0, 181, 10):
    radian = math.radians(degree)
    sin_value = math.sin(radian)
    print(f"sin({degree:3d}) = {sin_value:.2f}")
```

```python
from datetime import date

start_date = date(2019, 2, 14)
today = date.today()
passed_days = today - start_date

print(f"춘향이와 몽룡이의 연애 시작일 : {start_date.year}년 {start_date.month}월 {start_date.day}일")
print(f"연애 시작일로부터 경과한 날짜 : {passed_days.days}일")
```

```python
from datetime import date, timedelta

start_date = date(2019, 2, 14)

print(f"춘향이와 몽룡이의 연애 시작일 : {start_date.year}년 {start_date.month}월 {start_date.day}일")

days_list = [100, 200, 500, 1000]

for days in days_list:
    
    anniversary = start_date + timedelta(days=days-1)
    print(f"{days:,}일 기념일 : {anniversary.year}년 {anniversary.month}월 {anniversary.day}일")
```

```python
import time

def sum1to1000000():
    total = 0
    for i in range(1, 1000001):
        total += i
    return total


start = time.time()
sum1to1000000()
end = time.time()

print(f"1에서 1,000,000까지의 합을 구하는 시간 : {end - start:.4f}초")

```

```python
import time

def sum1to1000000():
    total = 0
    for i in range(1, 1000001):
        total += i
    return total

start_100 = time.time()
for _ in range(100):
    sum1to1000000()
end_100 = time.time()

print(f"1,000,000까지의 합을 100번 반복해서 구하는 시간 : {end_100 - start_100:.4f}초")
```

```python
import time
import math


def print_hello():
 
    text = "Hello Python!"

start = time.time()
for _ in range(100):
    print_hello()
end = time.time()
print(f"Hello Python! 출력을 100번 반복하는 시간 : {end - start:.4f}초")


def calc_factorial():
    math.factorial(10000)

start = time.time()
for _ in range(100):
    calc_factorial()
end = time.time()
print(f"10000!을 100번 반복해서 구하는 시간 : {end - start:.4f}초")


def sum_odd_cubes():
    total = sum(i ** 3 for i in range(1, 10001, 2))
    return total

start = time.time()
for _ in range(100):
    sum_odd_cubes()
end = time.time()
print(f"1에서 10000까지 홀수의 세제곱 더하기를 100번 반복하는 시간 : {end - start:.4f}초")


def sum_sin_values():
    total = sum(math.sin(math.radians(d)) for d in range(1, 361))
    return total

start = time.time()
for _ in range(100):
    sum_sin_values()
end = time.time()
print(f"1에서 360도까지 sin 값의 합을 100번 반복하는 시간 : {end - start:.4f}초")
```
